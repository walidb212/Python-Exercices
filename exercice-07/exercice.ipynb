```python
import seaborn as sns

titanic = sns.load_dataset("titanic")
```

```python
mineurs = titanic[titanic["age"] < 18]
print(f"Moins de 18 ans : {len(mineurs)}")
```

```python
top5_billets = titanic.sort_values("fare", ascending=False).head(5)
top5_billets[["fare", "class", "embark_town"]]
```

```python
print(f"Billet le plus cher : {titanic['fare'].max():.2f}")
```

```python
premiere_classe = titanic[titanic["pclass"] == 1]
femmes_1re = premiere_classe[premiere_classe["sex"] == "female"]
part_femmes_1re = len(femmes_1re) / len(premiere_classe) * 100
print(f"Femmes en 1re classe : {part_femmes_1re:.1f} %")
```

```python
hommes_3e = titanic[(titanic["sex"] == "male") & (titanic["pclass"] == 3)]

survie_femmes_1re = femmes_1re["survived"].mean() * 100
survie_hommes_3e = hommes_3e["survived"].mean() * 100
print(f"Survie femmes 1re : {survie_femmes_1re:.1f} %")
print(f"Survie hommes 3e : {survie_hommes_3e:.1f} %")
```
