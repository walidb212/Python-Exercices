```python
import pandas as pd

titanic = pd.read_csv("../data/titanic_propre.csv")
```

## 1. Femmes et hommes par classe

```python
pd.pivot_table(titanic, values="survived", index="class", columns="sex", aggfunc="mean").round(2)
```

Oui, dans les 3 classes les femmes ont plus survécu. En 1re : 97 % des femmes contre 37 % des hommes.

## 2. Les enfants

```python
titanic["tranche_age"] = pd.cut(
    titanic["age"], bins=[0, 12, 18, 60, 100],
    labels=["enfant", "ado", "adulte", "senior"],
)
titanic.groupby("tranche_age", observed=True)["survived"].mean().round(2)
```

Oui un peu : 58 % des enfants ont survécu contre 36 % des adultes.

## 3. Prix moyen par port

```python
titanic.groupby("embark_town")["fare"].mean().round(2)
```

```python
pd.crosstab(titanic["embark_town"], titanic["class"], normalize="index").round(2)
```

Cherbourg est le plus cher (60). C'est normal : la moitié des passagers de Cherbourg sont en 1re classe.

## 4. Voyager seul

```python
titanic.groupby("alone")["survived"].mean().round(2)
```

Voyager seul diminue les chances : 30 % de survie contre 51 % pour ceux qui ne sont pas seuls.
