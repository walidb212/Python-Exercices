```python
import pandas as pd

titanic = pd.read_csv("../data/titanic_propre.csv")

ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"],
})
```

```python
nb_avant = len(titanic)
titanic_pays = titanic.merge(ports, on="embark_town", how="left", validate="many_to_one")
assert len(titanic_pays) == nb_avant
print(nb_avant, len(titanic_pays))
```

```python
titanic_pays["pays"].value_counts()
```

```python
titanic_pays.groupby("pays")["survived"].mean().round(2)
```

```python
ports_sans_queenstown = ports[ports["embark_town"] != "Queenstown"]

jointure_inner = titanic.merge(ports_sans_queenstown, on="embark_town", how="inner")
jointure_left = titanic.merge(ports_sans_queenstown, on="embark_town", how="left")
print("inner :", len(jointure_inner))
print("left :", len(jointure_left))
print("pays manquants :", jointure_left["pays"].isna().sum())
```

En inner on perd les 77 passagers de Queenstown (812 lignes au lieu de 889) sans aucun message.

En left on garde les 889 lignes mais les 77 de Queenstown n'ont pas de pays.
