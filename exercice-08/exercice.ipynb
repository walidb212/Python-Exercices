```python
import seaborn as sns

titanic = sns.load_dataset("titanic")
titanic.isna().sum()
```

age : 177 manquants, je remplace par la médiane de la classe et du sexe

```python
mediane_age = titanic.groupby(["pclass", "sex"])["age"].transform("median")
titanic["age"] = titanic["age"].fillna(mediane_age)
```

embark_town : seulement 2 manquants, je supprime les lignes

```python
titanic = titanic.dropna(subset=["embark_town"])
print(len(titanic))
```

deck : 688 manquants sur 891 (77 %), je supprime la colonne

```python
titanic = titanic.drop(columns=["deck"])
```

```python
print(titanic.duplicated().sum())
titanic[titanic.duplicated(keep=False)].head()
```

Il n'y a pas de nom ni d'identifiant, donc 2 lignes pareilles peuvent être 2 vrais passagers. Je garde les doublons.

Colonnes en double : embarked = embark_town, pclass = class, alive = survived. Je les supprime.

```python
titanic = titanic.drop(columns=["embarked", "pclass", "alive"])
titanic["survived"] = titanic["survived"].astype(bool)
titanic.dtypes
```

```python
q1 = titanic["fare"].quantile(0.25)
q3 = titanic["fare"].quantile(0.75)
borne_haute = q3 + 1.5 * (q3 - q1)

billets_chers = titanic[titanic["fare"] > borne_haute]
print(len(billets_chers))
billets_chers["class"].value_counts()
```

Les billets chers sont presque tous en 1re classe, ce sont des vrais prix. Je les garde.

```python
titanic.to_csv("../data/titanic_propre.csv", index=False)
print(titanic.shape)
```
