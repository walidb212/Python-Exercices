```python
import seaborn as sns

titanic = sns.load_dataset("titanic")

print(titanic.shape)
print()

manquants = titanic.drop(columns="embarked").isna().sum()
for col, n in manquants[manquants > 0].items():
    print(f"{col:<12} {n:>3} manquants")
print()

for classe, n in titanic["class"].value_counts().items():
    print(f"{classe:<7} {n}")
```
