```python
import pandas as pd
import seaborn as sns

taxis = sns.load_dataset("taxis")
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
print(taxis.shape)
print(taxis["pickup"].min())
print(taxis["pickup"].max())
```

```python
courses_par_jour = taxis.set_index("pickup").resample("D")["total"].count()
ca_par_jour = taxis.set_index("pickup").resample("D")["total"].sum()
print(courses_par_jour.head())
print(ca_par_jour.head())
```

```python
taxis["heure"] = taxis["pickup"].dt.hour
taxis["heure"].value_counts().head(3)
```

L'heure la plus chargée c'est 18h.

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()
taxis.groupby("jour_semaine")["tip"].mean().sort_values(ascending=False).round(2)
```

Le mercredi a le pourboire moyen le plus élevé (2,09 $). Le samedi est le plus bas.

```python
taxis["duree"] = (taxis["dropoff"] - taxis["pickup"]).dt.total_seconds() / 60
taxis["duree"].describe()
```

```python
print("durées négatives :", (taxis["duree"] < 0).sum())
print("moins d'une minute :", (taxis["duree"] < 1).sum())
taxis[taxis["duree"] < 1][["duree", "distance", "fare"]].head()
```

Pas de durée négative et la plus longue fait 108 min, c'est possible.

Par contre 72 courses durent moins d'une minute, souvent avec 0 mile et 2,50 $ : sûrement des courses annulées. Je les supprime.

```python
taxis = taxis[taxis["duree"] >= 1]
print(len(taxis))
```
